## 1. Contexte général

La conduite automobile est une activité complexe qui mobilise plusieurs ressources cognitives : perception visuelle, attention soutenue, prise de décision, anticipation, contrôle moteur et gestion des imprévus.

Le papier CL-Drive s'intéresse donc à une question centrale :

> Peut-on estimer automatiquement la charge cognitive d'un conducteur à partir de signaux physiologiques et comportementaux mesurés pendant la conduite ?

#### Q1. Pourquoi est-il intéressant d'estimer la charge cognitive d'un conducteur ?

**Réponse :**

Une charge cognitive trop élevée dégrade les performances de conduite et augmente le risque d'accidents. Si un véhicule intelligent peut détecter automatiquement un niveau de charge cognitive dangereux, il peut avertir le conducteur ou adapter l'interface (réduire les distractions, ajuster les aides à la conduite) pour améliorer la sécurité routière.


## 2. Définition : charge cognitive

#### Q2. Comment définir la charge cognitive ?

**Réponse :**

La charge cognitive désigne la quantité de ressources mentales mobilisées pour accomplir une tâche. Plus une tâche est complexe, plus elle consomme de ressources cognitives. Elle se subdivise en charge intrinsèque (complexité inhérente de la tâche) et charge extrinsèque (ressources supplémentaires imposées par l'environnement, indépendantes de la tâche).

#### Q3. Dans le contexte de la conduite, de quels facteurs la charge cognitive peut-elle dépendre ?

**Réponse :**

- La complexité du scénario de conduite (nuit, neige, manœuvres difficiles)
- La réalisation de tâches secondaires (conversation, musique, téléphone)
- Les conditions environnementales (trafic, météo, luminosité)
- L'expérience et le niveau d'expertise du conducteur
- La fatigue et l'état émotionnel du conducteur

Dans le papier, la charge cognitive est évaluée à l'aide d'une échelle subjective appelée **PAAS**, allant de 1 à 9.

Les participants rapportent leur charge cognitive sur une échelle de 1 à 9.

#### Q5. Pourquoi parle-t-on ici d'une vérité terrain subjective ?

**Réponse :**

Parce que le score est auto-rapporté par le participant lui-même : c'est sa perception personnelle de sa charge cognitive, pas une mesure objective et physique. Deux individus dans la même situation peuvent rapporter des scores différents, et un même individu peut être inconsistant dans le temps. La mesure dépend donc de l'interprétation subjective de chacun.

## 5. Modalités mesurées

Le dataset CL-Drive est **multimodal**. Il contient plusieurs types de signaux enregistrés pendant la même tâche de conduite.

#### Q7. Compléter le tableau suivant.

| Modalité | Signification | Signal mesuré | Rôle possible |
|---|---|---|---|
| EEG | Électroencéphalogramme | Activité électrique cérébrale (4 canaux : AF7, AF8, TP9, TP10) | Refléter directement l'activité mentale et la charge cognitive |
| ECG | Électrocardiogramme | Activité électrique du cœur (variabilité du rythme cardiaque) | Indicateur du système nerveux autonome, corrélé à la charge mentale |
| EDA | Activité électrodermale | Conductance de la peau (réponse galvanique) | Indicateur du système nerveux sympathique, lié à l'arousal et au stress |
| Gaze | Suivi oculaire | Mouvements des yeux (saccades, fixations, diamètre pupillaire, clignements) | Refléter l'attention visuelle et la charge cognitive via la dilatation pupillaire |

Dans le projet, la modalité principale sera l’**EEG**. Les autres modalités serviront éventuellement d’extensions pour discuter de la fusion multimodale.

#### Q8. Pourquoi peut-il être intéressant de combiner EEG, ECG, EDA et Gaze pour estimer la charge cognitive ?

**Réponse :**

Chaque modalité capture un aspect différent de la réponse physiologique à la charge cognitive. L'EEG reflète directement l'activité cérébrale, l'ECG capture la réponse du système nerveux autonome, l'EDA mesure l'arousal sympathique, et le Gaze révèle l'attention visuelle. Aucune modalité seule n'est suffisamment robuste ou complète ; leur combinaison apporte des informations complémentaires et redondantes qui améliorent la précision de la classification.

## 7. Protocole expérimental

#### Q11. Quelle est la différence entre un **protocole expérimental** et un **dataset exploitable en machine learning** ?

**Réponse :**

Le protocole expérimental décrit comment les données ont été collectées : le déroulement des sessions, les consignes aux participants, les capteurs utilisés, les scénarios et leurs durées. Le dataset exploitable en ML est le résultat du protocole après nettoyage, prétraitement et mise en forme : des paires (fenêtre de signal, label) prêtes à être utilisées pour entraîner et évaluer un modèle. Le premier est un plan méthodologique, le second est une structure de données informatique.

#### Q12. Donner une description détaillée du protocole expérimental utilisé.

**Réponse :**

1. Pose des capteurs (EEG, ECG, EDA, Gaze) sur le participant
2. Enregistrement d'une baseline de 3 minutes (signal au repos)
3. Scénario d'orientation : conduite sur autoroute simple pour familiariser le participant avec le simulateur
4. Pour chacun des 9 scénarios principaux :

    - Repos de 2 minutes
    - Nouvelle baseline de 2 minutes
    - Scénario de conduite de 3 minutes (avec enregistrement continu)
    - Toutes les 10 secondes : signal audio invitant le participant à annoncer verbalement son score PAAS (noté par un expérimentateur)


5. Monitoring du niveau de SAS (Simulation Adaptation Syndrome) toutes les minutes sur une échelle de 1 à 9 ; arrêt immédiat si signes de malaise

#### Q13. Pourquoi les auteurs collectent-ils une baseline avant les scénarios ?

**Réponse :**

La baseline sert de référence individuelle au repos pour chaque participant. Elle permet de normaliser les signaux physiologiques par rapport à l'état de base de chaque sujet, ce qui réduit la variabilité inter-individuelle (chaque personne a des niveaux physiologiques différents au repos).

#### Q14. Pourquoi prévoir un scénario d’orientation avant les scénarios principaux ?

**Réponse :**

Pour que les participants se familiarisent avec le simulateur, comprennent les indicateurs de vitesse, le volant, les écrans, et s'habituent à l'environnement immersif avant de commencer les scénarios réels. Cela réduit aussi le SAS et assure que les données collectées reflètent la charge cognitive de la tâche et non la nouveauté du dispositif.

#### Q15. Pourquoi prévoir des périodes de repos entre les scénarios ?

### Réponse

Pour permettre au participant de récupérer et de revenir à un niveau de charge cognitive basal, éviter les effets de fatigue accumulée entre les scénarios, et réduire les risques de SAS. Cela garantit aussi que chaque scénario est réalisé dans des conditions comparables.

#### Q16. Pourquoi demander un score de charge cognitive toutes les 10 secondes plutôt qu’une seule fois à la fin du scénario ?

### Réponse

Un score unique en fin de scénario serait peu fiable (mémoire imparfaite, moyennage implicite). Des scores fréquents (toutes les 10 s) permettent de capturer les variations dynamiques de la charge cognitive au cours du scénario, d'obtenir 18 exemples labellisés par scénario, et d'entraîner des modèles plus précis. Les auteurs ont montré expérimentalement que 10 secondes est un bon compromis : plus court perturberait la conduite, plus long ferait oublier les expériences précédentes.

## 8. Scénarios de conduite

Le dataset contient 9 scénarios principaux, conçus pour induire différents niveaux de charge cognitive.

#### Q17. Compléter le tableau suivant.

| Scénario | Description synthétique |
|---:|---|
| 1 | Conduite sur autoroute de jour à 80 km/h |
| 2 | Conduite de nuit à 80 km/h |
| 3 | Conduite de nuit sur autoroute avec neige à 80 km/h |
| 4 | Défi balle de tennis : frapper la balle avec le pneu, précision et accélération |
| 5 | Défi slalom : passer entre des portes, précision et accélération |
| 6 | Défi passage étroit : naviguer entre des portes, précision et accélération |
| 7 | Défi virages à 90° : tourner à gauche et à droite, précision et accélération |
| 8 | Défi demi-tour : demi-tour en 3 points selon des instructions |
| 9 | Défi allée étroite : naviguer dans une allée très étroite, précision et accélération |

L'hypothèse générale est que les scénarios les plus complexes induisent une charge cognitive plus élevée.

#### Q18. Pourquoi utiliser plusieurs scénarios de conduite au lieu d'un seul scénario ?

**Réponse :**

Pour couvrir un spectre large de niveaux de charge cognitive (de faible à très élevée), assurer la variabilité des données d'entraînement, et éviter que les modèles ne surapprennent un seul type de situation. Plusieurs scénarios permettent aussi de vérifier que la classification fonctionne dans des conditions de conduite diverses.

#### Q19. Pourquoi peut-on s’attendre à une charge cognitive plus élevée dans les scénarios 8 et 9 ?

### Réponse

Le scénario 8 (demi-tour en 3 points dans une rue étroite) requiert une séquence précise de manœuvres sous contrainte spatiale forte. Le scénario 9 (allée très étroite) impose une précision latérale extrême à chaque instant. Ces deux situations sollicitent fortement l'attention, la planification et le contrôle moteur simultanément. La heatmap de la Figure 4 du papier confirme que les scores PAAS les plus élevés sont concentrés sur ces deux scénarios.

#### Q20. La complexité du scénario suffit-elle à définir le label de charge cognitive ?

### Réponse

Non. La complexité du scénario est un inducteur de charge cognitive, mais le label reste le score PAAS auto-rapporté par le participant. Deux sujets dans le même scénario peuvent rapporter des scores différents selon leur niveau d'expertise, leur état de fatigue, leur personnalité ou leur seuil de sensibilité. Les auteurs utilisent donc les scores subjectifs comme vérité terrain et non la complexité du scénario directement.

## 9. Échantillonnage et segmentation temporelle

Les signaux EEG sont continus. Pour entraîner un modèle, il faut transformer ce signal continu en exemples individuels.

Dans CL-Drive, les scores PAAS de charge cognitive sont collectés toutes les 10 secondes.

Il est donc naturel de découper les signaux physiologiques en fenêtres de 10 secondes :

$$
X_i = \text{segment de 10 s}
$$

$$
y_i = \text{score PAAS associé}
$$

On obtient ainsi le problème suivant :

Le principe d’association segment-label est le suivant :

| Temps | Segment EEG | Score PAAS |
|---|---|---|
| 0–10 s | segment 1 | score 1 |
| 10–20 s | segment 2 | score 2 |
| 20–30 s | segment 3 | score 3 |

256 Hz (256 échantillons par seconde par canal).

**Réponse :**

### Réponse

## 10. Formulation du problème d'apprentissage automatique

À partir du dataset CL-Drive, on peut formuler plusieurs problèmes de classification.

### Classification binaire

| Scores PAAS | Classe |
|---|---|
| 1 à 4 | charge faible |
| 5 à 9 | charge élevée |

### Classification ternaire

| Scores PAAS | Classe |
|---|---|
| 1 à 3 | charge faible |
| 4 à 6 | charge moyenne |
| 7 à 9 | charge élevée |

Dans le projet, on commencera par la classification binaire.

#### Q27. Pourquoi commencer par une classification binaire plutôt qu’une classification ternaire ?

### Réponse

La classification binaire est plus simple : 2 classes seulement (charge faible / charge élevée), frontière de décision moins complexe, moins de risque de déséquilibre de classes, résultats plus faciles à interpréter et à valider. C'est une étape de validation préalable avant d'aborder la classification ternaire (plus difficile et moins bien résolue, comme le confirment les résultats du papier).

#### Q28. Dans une classification binaire, pourquoi le score PAAS 5 est-il généralement placé dans la classe « charge élevée » dans ce papier ?

### Réponse

Le score 5 correspond à "ni faible ni élevée", donc un état neutre. Dans un contexte de sécurité routière, il est plus prudent d'assimiler cet état ambigu à la classe à risque (charge élevée) plutôt qu'à la classe sans risque. Le seuil de coupure à 4/5 reflète ainsi un principe conservateur : mieux vaut déclencher une alerte inutile que de manquer une surcharge cognitive réelle.

#### Q29. Quel est le lien entre ce papier et un système d'interface adaptative ?

**Réponse :**

Un tel système embarqué pourrait, en temps réel, classifier la charge cognitive du conducteur à partir des signaux physiologiques, et adapter automatiquement l'interface véhicule en conséquence : réduire les notifications, simplifier l'affichage, déclencher une alerte sonore ou visuelle, ou même proposer une pause. CL-Drive fournit exactement les données et les modèles de base pour développer et évaluer de tels systèmes.

## 9. Visualisation des signaux EEG

Avant tout prétraitement ou apprentissage automatique, il est indispensable de visualiser les signaux EEG.

La visualisation permet de :

- vérifier que les données sont correctement chargées ;
- observer les amplitudes ;
- repérer les valeurs aberrantes ;
- détecter les ruptures ou données manquantes ;
- comparer les canaux ;
- repérer des artefacts évidents ;
- comparer qualitativement des segments de faible et forte charge cognitive.

### Algorithme conceptuel pour visualiser un segment EEG

Pour visualiser un segment EEG de 10 secondes :

1. choisir un sujet ;
2. choisir un scénario ;
3. choisir un segment ;
4. récupérer les 4 canaux EEG ;
5. calculer le nombre d’échantillons : $N = f_s \times T$ ;
6. construire l’axe temporel : $t[n] = n/f_s$ ;
7. tracer les quatre canaux en fonction du temps ;
8. ajouter le score PAAS associé au titre du graphique ;
9. commenter la qualité du signal.

#### Q30. Que faut-il regarder en premier lorsqu’on affiche un signal EEG brut ?

### Réponse

- L'amplitude globale (ordre de grandeur en µV, typiquement 10–100 µV)
- La présence de valeurs aberrantes ou de saturations
- Les ruptures ou données manquantes (gaps)
- Les artefacts évidents (clignements, mouvements musculaires)
- La cohérence temporelle (signal continu et plausible)

#### Q31. Pourquoi afficher les quatre canaux EEG plutôt qu’un seul ?

### Réponse

Pour détecter des problèmes de contact spécifiques à un canal, observer des différences spatiales entre régions frontales et temporales, repérer des artefacts locaux, et vérifier la qualité globale de l'acquisition. Un seul canal pourrait être défectueux sans que les autres le soient.

#### Q32. Pourquoi comparer visuellement un segment de faible charge et un segment de forte charge cognitive ?

### Réponse

Pour vérifier intuitivement que le signal EEG contient bien de l'information discriminante, repérer des différences qualitatives (amplitude, fréquence dominante), et valider que les labels sont cohérents avec les signaux. Si aucune différence visuelle n'est observable, cela questionne la qualité des données ou des labels.

#### Q33. Pourquoi l'utilisation d'un simulateur est-elle à la fois un avantage et une limite ?

**Réponse :**

Avantages : contrôle total des conditions expérimentales (météo, trafic, heure du jour), sécurité totale pour les participants, reproduction exacte des mêmes scénarios pour tous les sujets, collecte de signaux sans danger.
Limites : risque de Simulation Adaptation Syndrome (SAS) causant malaises et arrêts prématurés, graphismes et sensations haptiques qui ne reproduisent pas parfaitement la réalité, comportements des conducteurs potentiellement différents en situation simulée vs réelle.

Le papier CL-Drive présente un nouveau dataset multimodal pour l'estimation automatique de la charge cognitive des conducteurs. Le problème central est de savoir si des signaux physiologiques mesurés en temps réel pendant la conduite permettent de prédire le niveau de charge cognitive d'un conducteur. Les données ont été collectées sur 21 participants conduisant dans un simulateur de véhicule immersif, répartis sur 9 scénarios de complexité croissante (de la conduite autoroutière de jour aux manœuvres délicates en allée étroite). Quatre modalités sont enregistrées simultanément : EEG (activité cérébrale), ECG (rythme cardiaque), EDA (conductance cutanée) et Gaze (suivi oculaire). Les labels de vérité terrain sont des scores de charge cognitive subjectifs (échelle PAAS de 1 à 9) auto-rapportés par les participants toutes les 10 secondes. Ces scores sont ensuite regroupés en classes binaires (faible/élevée) ou ternaires (faible/moyenne/élevée) pour formuler le problème comme une tâche de classification supervisée. Les auteurs évaluent plusieurs modèles (Random Forest, XGBoost, VGG, ResNet) selon deux protocoles de validation : 10-fold et Leave-One-Subject-Out. Les meilleurs résultats atteignent 83 % de précision en binaire (10-fold, XGBoost). Ce dataset ouvre la voie au développement de systèmes embarqués d'interfaces adaptatives capables de détecter une surcharge cognitive et d'alerter ou d'assister le conducteur en conséquence.
